```python
# Ex.2

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"Hello! I am {self.name}, {self.age} years old"

# Creating the object with the respective values
p1 = Person("Ana", 35)
p2 = Person("Paulo", 26)

print(p1.introduce())
```

```python
# Ex.3

class Student:
    def __init__(self, name, college, course, semester):
        self.name = name
        self.college = college
        self.course = course
        self.semester = semester

    def introduce_student(self):
        text = f"Hello, my name is {self.name}. I study {self.course} at {self.college} and currently I am in my {self.semester}th semester."
        print(text)

my_student = Student("Yago", "FIAP", "Engineering", 4)

my_student.introduce_student()
```

```python
# Ex.4 

class Rectangle:
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return self.base * self.height

    def perimeter(self):
        return 2 * (self.base + self.height)

r1 = Rectangle(float(input("Base: ")), float(input("Height: ")))
print(f"The area of r1: {r1.area()} - Perimeter: {r1.perimeter()}")
```

```python
# Ex.5

class Person:
    def __init__(self, name, weight, height):
        self.name = name
        self.weight = weight
        self.height = height

    def calculate_bmi(self):
        bmi = self.weight / (self.height ** 2)
        return bmi

    def present_result(self):
        # Alteração feita aqui!
        bmi = self.calculate_bmi()
        print(f"The BMI of {self.name} is: {bmi:.2f}")

print("--- BMI Results ---")

person1 = Person("Ana", 65.0, 1.68)
person2 = Person("Carlos", 85.5, 1.75)
person3 = Person("Beatriz", 58.0, 1.60)

person1.present_result()
person2.present_result()
person3.present_result()
```

```python
# Ex.6

class Investment:
    def __init__(self, name, capital, rate, time):
        self.name = name
        self.capital = capital
        self.rate = rate
        self.time = time

    def calculate_amount(self):
        amount = self.capital * (1 + self.rate / 100) ** self.time
        return amount

    def calculate_profit(self):
        amount = self.calculate_amount()
        profit = amount - self.capital
        return profit

    def present_result(self):
        amount = self.calculate_amount()
        profit = self.calculate_profit()
        print(f"--- {self.name} ---")
        print(f"Amount: ${amount:.2f}")
        print(f"Profit: ${profit:.2f}\n")

inv1 = Investment("CDB", 1000.0, 1.0, 12)
inv2 = Investment("Treasury Bonds", 2000.0, 0.85, 24)
inv3 = Investment("Fund Y", 5000.0, 1.2, 6)

inv1.present_result()
inv2.present_result()
inv3.present_result()
```

```python
# Ex.7

class Book:
    # 1. Receiving the book information
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        # Condition: long_read if pages > 300
        self.long_read = self.pages > 300

    # 2. Method to build the catalog card
    def catalog_card(self):
        # Creating the string with the expected output format
        card = f"'{self.title}', by {self.author} ({self.pages} p.)"
        
        # Adding a visual warning if it is a long read
        if self.long_read:
            card += " [Warning: Long Read]"
            
        print(card)

# --- Testing with the exact data from the image ---
book1 = Book("Dom Casmurro", "Machado de Assis", 256)
book2 = Book("War and Peace", "Leo Tolstoy", 1200)
book3 = Book("The Little Prince", "Antoine de Saint-Exupéry", 96)

print("--- Catalog Cards ---")
book1.catalog_card()
book2.catalog_card()
book3.catalog_card()
```
